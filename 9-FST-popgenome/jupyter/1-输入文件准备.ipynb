```python
var_list = ['Pepticulcer-消化性溃疡', 'Atrophy-萎缩',
            'Atrophyscore*-萎缩评分', 'Intestinalmetaplasia-肠化生',
            'AML-阿莫西林', 'TE-四环素','MTZ-甲硝唑',
            'LEV-左氧氟沙星', 'CLR-克拉霉素', 'RIF-利福平']
```

```python
import os 
for var in var_list:
    os.makedirs(f'../../output/临床指标/{var}/',exist_ok=True)
    df_clinic_temp  = df_clinic.loc[:,['ID',var]].rename(columns={'ID':'strain',
                                                                 var:'fs_pop'})
    df_clinic_temp['fs_pop'] = df_clinic_temp['fs_pop'].map({0:'hsp1',1:'hsp2'})
    df_clinic_temp.loc[:,['strain','fs_pop']].dropna().to_csv(f'../../output/临床指标/{var}/META.csv',index=False)
```
