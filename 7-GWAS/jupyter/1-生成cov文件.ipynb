```python
df_PCA = pd.read_csv('/mnt/c/Users/Administrator/Desktop/data/Han_SNP4.eigenvec',sep=' ',header=None)
df_PCA
```

```python
# 重命名df_PCA
df_PCA.columns = ['FID', 'IID'] + [f'PC{i+1}' for i in range(df_PCA.shape[1] - 2)]
df_PCA
```

```python
df_out = pd.read_csv('/mnt/c/Users/Administrator/Desktop/data/Han_SNP4.out',sep=r'\s+')
mask = df_out['p-value'] <= 0.05
# 在原 df_out 上添加新列，仅针对满足条件的行
df_out.loc[mask, 'PC_value'] = 'PC' + df_out.loc[mask, '#N'].astype(str)
# 如果你只想保留显著的部分，再切片
df_out_sig = df_out.loc[mask]
df_out_sig
```

```python
# 将符合要求的PC值（即df_out_sig中的PC_value列）从df_PCA中对应的列提取出来
pc_columns = df_out_sig['PC_value'].unique()
df_PCAS = df_PCA[['FID', 'IID'] + list(pc_columns)]
df_PCAS
```

```python
# 打印
print('共提取了{}个样本的{}个主成分'.format(df_PCAS.shape[0], len(pc_columns)))
print('分别是:', ', '.join(pc_columns))
```
