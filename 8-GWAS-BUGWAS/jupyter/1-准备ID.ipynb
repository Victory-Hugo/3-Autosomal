```python
import pandas as pd
import os
```

```python
Cinic_file = '/mnt/f/OneDrive/文档（科研）/P0_幽门螺旋杆菌/基础信息表/临床表型.csv'
GWAS_CONF_DIR = '/mnt/d/幽门螺旋杆菌/Script/分析结果/8-临床结局相关/data/GWAS'
TARGET_PHENO = 'Atrophy-萎缩'
```

```python
df_Clinical = pd.read_csv(Cinic_file)
df_Clinical
```

```python
# 获取列名
clinical_columns = df_Clinical.columns.tolist()
clinical_columns
```

```python
os.makedirs(f'{GWAS_CONF_DIR}/{TARGET_PHENO}', exist_ok=True)
df_temp = df_Clinical.loc[:,['ID',TARGET_PHENO]].dropna().rename(columns={'ID':'ID', TARGET_PHENO:'pheno'})
```

```python
# 保存ID列表和表型数据
# 将pheno列保存为整数
df_temp['pheno'] = df_temp['pheno'].astype(int)
df_temp.loc[:,['ID']].to_csv(f'{GWAS_CONF_DIR}/{TARGET_PHENO}/list.txt', index=False, header=False ,float_format='%.0f', sep='\t')
df_temp.to_csv(f'{GWAS_CONF_DIR}/{TARGET_PHENO}/pheno.txt', index=False, sep='\t')
```
