```python
import pandas as pd
```

```python
Q_FILE = '/mnt/d/幽门螺旋杆菌/Script/分析结果/ADMIXTURE/data/WGS.aln.fasta.CDS_地理东亚及周边_3378_filtered.6.Q'
FAM_FILE = '/mnt/d/幽门螺旋杆菌/Script/分析结果/ADMIXTURE/data/WGS.aln.fasta.CDS_地理东亚及周边_3378_filtered.fam'
META_FILE = '/mnt/f/OneDrive/文档（科研）/P0_幽门螺旋杆菌/基础信息表/HP数据收集2.xlsx'
```

```python
df_Q = pd.read_csv(Q_FILE, sep=' ', header=None)
df_Q.columns = [f'K{i+1}' for i in range(df_Q.shape[1])]
df_fam = pd.read_csv(FAM_FILE, sep='\t', header=None).loc[:,[0,1]]
df_fam.columns = ['FID', 'ID']
df_fam
```

```python
df_Q
```

```python
# 将Q矩阵与fam文件合并
df_Q_fam = pd.concat([df_fam, df_Q], axis=1)
df_Q_fam
```

```python
df_meta = pd.read_excel(META_FILE, sheet_name='HP数据收集').loc[:,['ID','Latitude','Longitude']]
df_meta.head()
```

```python
# 合并数据
df_combined = pd.merge(df_Q_fam, df_meta, left_on='ID', right_on='ID', how='left')
df_combined.head()
```
